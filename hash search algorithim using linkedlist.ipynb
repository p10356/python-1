```python
global db
db={}
```

```python
def userinput():
    while True:
        try:
            val = int(input("Enter a value"))
            break
        except (ValueError):
            print("Enter number only")
    return val
```

```python
def insert():
    val=userinput()
    hashkey=int((val%0.95)*40)
    if hashkey not in db:
        db[hashkey]=[(int(0),val)]
    else:
        c=0
        for i in db[hashkey]:
            if i[0]==int(0):
                temp=list(i)
                temp[0]=id(val)
                db[hashkey][c]=tuple(temp)
            c+=1
        db[hashkey].append((int(0),val))
```

```python
def delete():
    val=userinput()
    hashkey=int((val%0.95)*40)
    if hashkey not in db:
        print("Value Not Found as stored with Hash function")
    else:
        index=0
        for i in range(len(db[hashkey])):
            if db[hashkey][i][1] == val:
                index=i
        if index == 0:
            if len(db[hashkey]) == 1:
                db.pop(hashkey)
            else:
                db[hashkey].pop(index)
        else:
            _id=db[hashkey][index][0]
            _val=db[hashkey][index-1][1]
            db[hashkey][index-1]=(_id,_val)
            db[hashkey].pop(index)
```

```python
def search():
    val=userinput()
    hashkey=int((val%0.95)*40)
    if hashkey in db:
        for i in db[hashkey]:
            if i[1] == val:
                print("Data Found")
                break
    else:
        print("Data Not Found")
```

```python
def display():
    print("-"*30)
    for key , value in db.items():
        print(key,":")
        l=len(value)
        lc=1
        for i in value:
            print(i[1],end="")
            if l > lc:
                print(end=",")
                lc+=1
        print()
        print("-"*30)
```

```python
def main():
    while True:
        choice=input('''
Enter 
i for Insert
d for Delete
s for Search
l for list of data
x for Exit
: ''')
        print()
        if choice.lower() == 'i':
            insert()
        elif choice.lower() == 'd':
            delete()
        elif choice.lower() == 's':
            search()
        elif choice.lower() == 'l':
            display()            
        elif choice.lower() == 'x':
            break
        else:
            continue
    
```

```python
main()
```
